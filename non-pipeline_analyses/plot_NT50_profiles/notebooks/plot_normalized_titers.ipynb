# Normalize neutralization titers to each participant's geometric mean titer and plot

```python
# Define inputs
datadir = '../data'
resultsdir = '../results'
os.makedirs(datadir, exist_ok = True)
os.makedirs(resultsdir, exist_ok = True)
```

## Calculate normalized titers

```python
# Get all titers

# Define SCH titers
SCH_titers = (pd.read_csv('../../../results/aggregated_titers/titers_SCH.csv')
             .assign(
                 barcode = lambda x: x['serum'].str.split('_').str[2],)
             )

# Define Penn titers
titers_PennVaccineCohort = (pd.read_csv('../../../results/aggregated_titers/titers_PennVaccineCohort.csv')
                            .assign(
                                barcode = lambda x: x['serum'].str.split('_').str[2],
                                timepoint = lambda x: 'd' + x['serum'].str.split('d').str[1])
                           )

# Define Australian MA22 vaccine cohort titers
Australia_MA22_titers = (pd.read_csv('../../../results/aggregated_titers/titers_AusVaccineCohort.csv')
                         .assign(
                             barcode = lambda x: x['serum'].str.split('_').str[1],
                             timepoint = lambda x: x['serum'].str.split('_').str[2])
                        )

# Define sera metadata
SCH_metadata = pd.read_csv('../../../data/sera_metadata/metadata_SCH.csv')
AusVaccineCohort_metadata = (pd.read_csv('../../../data/sera_metadata/metadata_AusVaccineCohort.csv')
                             .rename(columns = {'Bloom_lab_serum_ID': 'serum'}))
# Reformatting the Penn metadata
day0 = (pd.read_csv('../../../data/sera_metadata/metadata_PennVaccineCohort.csv')
        .drop(['Bloom_lab_ID_d28'], axis=1)
        .rename(columns = {'Bloom_lab_ID_d0': 'serum'}))

day28 = (pd.read_csv('../../../data/sera_metadata/metadata_PennVaccineCohort.csv')
         .drop(['Bloom_lab_ID_d0'], axis=1)
         .rename(columns = {'Bloom_lab_ID_d28': 'serum'}))
PennVaccineCohort_metadata = pd.concat([day0,day28])
# Concatenate all metadata
all_metadata = pd.concat([SCH_metadata,PennVaccineCohort_metadata, AusVaccineCohort_metadata])


# Concatenate all titers into one dataframe
all_titers = pd.concat([
    SCH_titers,
    titers_PennVaccineCohort,
    Australia_MA22_titers
])

# Merge all titers and metadata on 'serum' key
all_titers = all_titers.merge(all_metadata, on = 'serum')

# Add a slightly more informative 'group_detail' column that incorporates pre- or post-vax info
all_titers['group_detail'] = np.where(all_titers['group'] == 'SCH', 
                                         all_titers['group'],  # If the value matches the string
                                         all_titers['group'].astype(str) + '_' + all_titers['timepoint'])  # Otherwise, combination of col2 and col3
```

```python
# Define virus order
viral_plot_order = pd.read_csv('../../../data/H3N2library_2023-2024_strain_order.csv')
virus_order = [v for v in viral_plot_order.strain]

# Define vaccine strains
vaccine_strains = []
with open('../data/vaccine_strains.csv') as f:
    for line in f:
        line = line.strip('\n')
        if 'strain' not in line:
            vaccine_strains.append(line)

# Define separate list where Massachusetts/18/2022 is reclassified as a 2023-circulating strains
vaccine_strains_no_Massachusetts = [item for item in vaccine_strains if item != 'A/Massachusetts/18/2022']
```

```python
# Get normalized titer, where each individual is normalized to their geometric mean titer

all_titers_2023circulating_only = (all_titers[~all_titers['virus'].isin(vaccine_strains_no_Massachusetts)]
                                   .reset_index(drop=True))

# Initialize empty list of normlized titers and gmeans
normalized_titers = []

# Get normlized titers
for serum in all_titers_2023circulating_only.serum.unique():
    temp_df = all_titers_2023circulating_only.query(f'serum == "{serum}"').copy()

    # Get geometric mean of all titers for that serum
    serum_gmean = gmean(temp_df.titer)

    temp_df['serum_gmean'] = serum_gmean
    temp_df['normalized_titer'] = temp_df['titer'] / serum_gmean
 
    normalized_titers.extend(temp_df[['serum', 'virus', 'serum_gmean', 'normalized_titer']].to_numpy().tolist())
    
all_titers_2023circulating_only = all_titers_2023circulating_only.merge(
    pd.DataFrame(normalized_titers, columns = ['serum', 'virus', 'serum_gmean', 'normalized_titer']),
    on = ['serum', 'virus']
)

all_titers_2023circulating_only['year_of_birth'] = all_titers_2023circulating_only['year_of_birth'].astype(float)
```

```python
# Add age range
age_range_list = []
for year in all_titers_2023circulating_only.year_of_birth:
    if year > 2020:
        age_range_list.append('born 2021-2023 (n=7)')
    elif year <= 2020 and year >= 2015:
        age_range_list.append('born 2015-2020 (n=25)')
    elif year < 2015 and year >= 2009:
        age_range_list.append('born 2009-2014 (n=24)')

    # Adults
    elif year < 2003 and year >= 1990 :
        age_range_list.append('born 1990-2002 (n=12)')
    elif year < 1990 and year >= 1980 :
        age_range_list.append('born 1980-1989 (n=11)')
    elif year < 1980 and year >= 1970 :
        age_range_list.append('born 1970-1979 (n=6)')
    elif year < 1970:
        age_range_list.append('born <1970 (n=10)')
                
    else:
        age_range_list.append('adults')


all_titers_2023circulating_only['age_range'] = (age_range_list)
```

```python
# Plot each cohort (SCH, adults pre, adults post, etc) of normalized titers

def plot_normalized_sera(data, facet_by, height=280,y_label=['normalized', 'neutralization titer']):
    """ Function for plotting individual sera in different panels """

    color_scheme = alt.Color('group', 
                             title = 'Cohort', 
                             sort=['SCH','PennVaccineCohort'],
                             legend=None
                            ).scale(range=color_palette)
    titer_range = [0, 2]
    titleFontSize=19
    labelFontSize=19
    lineOpacity = 0.18
    lineSize = 2.8
    markerOpacity = 0.9
    markerSize = 160
    width = 1300
    # height = 280
    

    line = (alt.Chart(data, width = width,height=height)
            .mark_line(size = lineSize, point = False, opacity = lineOpacity)
            .encode(
                alt.X('virus', 
                          axis = alt.Axis(grid=False, titleFontSize=titleFontSize, labelFontSize=labelFontSize,
                                          title = None,labelLimit = 1000, labelAlign = 'right',
                                          # labelColor = all_vacc_label_colors,labelFontWeight = all_vacc_label_weights,
                                         ),             
                          sort = virus_order
                         ),
                alt.Y('normalized_titer', 
                          scale =alt.Scale(type='log',
                                           # domain=titer_range, 
                                           nice=False), 
                          axis=alt.Axis(grid=False, titleFontSize=titleFontSize, labelFontSize=labelFontSize, title=y_label)
                     ),
                detail = 'serum',
                color = color_scheme,
                tooltip=['serum']
                # order = 'virus_order'
            )
           )
    
    points = (alt.Chart(data, width=width)
              .mark_point(size = markerSize,
                          stroke = 'black',
                          strokeWidth = 2.2,
                          filled=True, 
                          opacity=markerOpacity)
              .encode(
                  alt.X('virus', sort = virus_order),
                  alt.Y('median(normalized_titer)'),
                  color = color_scheme,
              )
             )
        
    layered = (alt.layer(line, points)
               .facet(
                   row = alt.Row(f'{facet_by}:N', 
                                 sort = ['SCH','PennVaccineCohort'], 
                                 title = None,
                                 # header=None
                                ),
                   config = alt.Config(
                            legend = alt.LegendConfig(titleFontSize=titleFontSize, labelFontSize = labelFontSize,
                                strokeColor='gray',padding=10,cornerRadius=10,
                                labelLimit = 1000 # Let legend labels be as long as they want
                            )
                           )
               ).configure_header(
                       # labels=False, # Removing labels for pretty versions of figure, comment out to see labels
                                  labelFontSize=labelFontSize,labelFontWeight='bold',
                                  labelOrient='right', labelAngle=0, labelAnchor='start', labelAlign='right')
    )


    return layered
```

## Correlate normalized titers and partipant ages

```python
def get_correlations_titers_vs_age(df, unique_sera):
    # Initialize empty list for all comparisons
    titer_vs_age_comarisons = []

    for sera1, sera2 in combinations(unique_sera, 2):
        # print(f"Comparing {sera1} vs {sera2}")
        sera_comparison = '_'.join([sera1, sera2])
    
        # Get age difference between sera
        sera1_age, sera2_age = df.loc[df['serum'] == f'{sera1}', 'age'].iloc[0], df.loc[df['serum'] == f'{sera2}', 'age'].iloc[0]
        age_difference = abs(sera1_age - sera2_age)
        
        # Narrow df and pivot values
        temp_df = df[df['serum'].isin([sera1, sera2])]
        pivot_df = temp_df.pivot_table(
            index=['virus',],
            columns='serum',
            values='normalized_titer'
        ).reset_index().dropna()
    
        
        # Define variables
        titer1, titer2 = pivot_df[[f'{sera1}']], pivot_df[[f'{sera2}']]
    
        if len(titer1) > 0 and len(titer2) > 0:
            # Initiate linear regression model
            model = LinearRegression()    
            # Fit regression model
            model.fit(titer1, titer2)  
            # Calculate R-squared of regression model
            r_squared = model.score(titer1, titer2)
    
            # Also note the two cohorts being compared
            group1 = sera1.split('_')[0]
            group2 = sera2.split('_')[0]
            cohort_comparison = '_'.join(sorted(set([group1, group2])))
    
            # Save R2, age difference, and string of cohort-cohort
            titer_vs_age_comarisons.append([
                sera_comparison,
                r_squared,
                age_difference,
                cohort_comparison
            ])
    
        
    titer_vs_age_df = pd.DataFrame(titer_vs_age_comarisons,
                 columns = [
                     'sera_comparison',
                     'r_squared',
                     'age_difference',
                     'cohort_comparison'
                 ])

    return titer_vs_age_df
```

```python
# Calculate correlation coeffcients and difference in ages for all combos and cohorts
timepoints_to_exclude = ['d28', 'postvax']

df = all_titers_2023circulating_only
df = df[~df['timepoint'].isin(timepoints_to_exclude)]
df = df.query('serum != "SCH23_y2021_s057"')
unique_sera = df.serum.unique()
df = df.copy() 

# Add age column
# Fill missing ages using year_of_birth
df['age'] = df['age'].fillna(2023 - df['year_of_birth'])
df['age'] = df['age'].round().astype('Int64')  # 'Int64' supports NA values

titer_vs_age_df = get_correlations_titers_vs_age(df, unique_sera)

titer_vs_age_df.cohort_comparison.unique()

```

```python
# Plot correlation coefficients versus difference in ages
cohorts_to_plot = ['SCH23', 'PENN23', 'PENN23_SCH23']
longer_label_dict = {'SCH23': 'children (n=56)', 
                     'PENN23': 'pre-vaccination adults (n=39)',
                     'PENN23_SCH23': 'adults and children (n=95)'}
df = (titer_vs_age_df
      [titer_vs_age_df['cohort_comparison'].isin(cohorts_to_plot)]
      .replace(longer_label_dict)
     )

# Configure simple scatter
fill = True
opacity = 0.15
stroke = 'black'
strokeWidth = 1.2
markSize = 120
titleFontSize = 16
labelFontSize = 16
RtextLocation = 90
width = 200
height = width
_range = [60, 30000]


charts = []

for cohort in [longer_label_dict.get(item, item) for item in cohorts_to_plot]:
    sub_df = df[df['cohort_comparison'] == cohort]

    # Fit R²
    if len(sub_df) > 1:
        model = LinearRegression()
        x = sub_df[['age_difference']]
        y = sub_df[['r_squared']]
        # Implement scaling
        scaler = StandardScaler()
        x_scaled = scaler.fit_transform(x)
        y_scaled = scaler.fit_transform(y)

        # Fit model and calculate R2
        model.fit(x_scaled, y_scaled.ravel())
        # r2 = model.score(x_scaled, y_scaled)
        # r2_string = f"R² = {r2:.2f}"

        # Predict
        y_pred = model.predict(x_scaled)
        # Compute R (Pearson correlation between predicted and actual values)
        print(f"Slope: {model.coef_[0]:.2f}")
        print(f"R: {np.corrcoef(y_scaled.ravel(), y_pred.ravel())[0, 1]}")
        r_string = f"R = {model.coef_[0]:.2f}"
    
    else:
        r_string = ""

    # Determine annotation position
    x_pos = sub_df['age_difference'].max() * 0.9
    y_pos = sub_df['r_squared'].max() * 0.9

    # Scatter chart
    scatter = alt.Chart(sub_df).mark_circle(
        size=markSize,
        stroke=stroke,
        strokeWidth=strokeWidth,
        filled=fill,
        opacity=opacity
    ).encode(
        x=alt.X('age_difference:Q',
                title=['age difference', 'between pairs of sera'],
                axis=alt.Axis(grid=False, titleFontSize=titleFontSize, labelFontSize=labelFontSize, titleFontWeight='normal')),
        y=alt.Y('r_squared:Q',
                title=['correlation coefficient (R2)', 'between pairs of sera'],
                scale = alt.Scale(domain=[0,1]),
                axis=alt.Axis(grid=False, titleFontSize=titleFontSize, labelFontSize=labelFontSize, titleFontWeight='normal')),
        tooltip=['sera_comparison', 'r_squared']
    ).properties(
        width=width,
        height=height,
        title=alt.TitleParams(text=cohort, fontSize=titleFontSize, anchor='middle', fontWeight='normal')
    )

    # Text annotation
    text = alt.Chart(pd.DataFrame({
        'label': [r_string]
    })).mark_text(
        align='right',
        baseline='top',
        dx=RtextLocation, dy=-RtextLocation,
        fontSize=labelFontSize,
        color='black'
    ).encode(
        text='label:N'
    )

    # Combine and append
    charts.append(scatter + text)

# --- Concatenate all charts horizontally ---
final_plot = alt.concat(*charts, columns=3).configure_axis(grid=False).properties(
    title=alt.TitleParams(
        text='titer correlations vs age difference among 2023-circulating strains', 
        anchor='middle', 
        fontSize=18,
    )
)

final_plot

# Save plot
outfile = os.path.join(resultsdir, 'titer_vs_age_circulating.svg')
final_plot.save(outfile, width = 1800)

# Display plot
final_plot
```

## Correlate normalized titers and strain differences

```python
ectodomain_distances = pd.read_csv('../../library_design/results/2023-2024_H3_library_protein_HA_ectodomain_distances.csv').set_index('Unnamed: 0')
ha1_distances = pd.read_csv('../../library_design/results/2023-2024_H3_library_protein_HA1_distances.csv').set_index('Unnamed: 0')
```

```python
def get_correlations_titers_vs_distance(df, unique_viruses):
    # Initialize empty list for all comparisons
    titer_vs_distance_comparisons = []

    for virus1, virus2 in combinations(unique_viruses, 2):
        # print(f"Comparing {virus1} vs {virus2}")
        virus_comparison = '_'.join([virus1, virus2])
        
        # Get distances
        ectodomain_dist = ectodomain_distances.loc[virus1, virus2]
        HA1_dist = ha1_distances.loc[virus1, virus2]
        # print(virus_comparison, ectodomain_dist, HA1_dist)
        
        # Subset df to just these two viruses
        temp_df = df[df['virus'].isin([virus1, virus2])]
        pivot_df = temp_df.pivot_table(
            index=['serum', 'group'],
            columns='virus',
            values='normalized_titer'
        ).reset_index().dropna()

        if pivot_df.empty:
            continue

        # Get unique group names
        groups = pivot_df['group'].unique()

        # Loop through all non-empty combinations of groups
        for r in range(1, len(groups) + 1):
            for combo in combinations(groups, r):
                combo_name = '_'.join(sorted(combo))
                sub_pivot = pivot_df[pivot_df['group'].isin(combo)]

                if sub_pivot.empty:
                    continue

                titer1 = sub_pivot[[virus1]]
                titer2 = sub_pivot[[virus2]]

                if len(titer1) > 0 and len(titer2) > 0:
                    model = LinearRegression()
                    model.fit(titer1, titer2)
                    r_squared = model.score(titer1, titer2)

                    titer_vs_distance_comparisons.append([
                        virus_comparison,
                        r_squared,
                        ectodomain_dist,
                        HA1_dist,
                        combo_name  # record specific group combo
                    ])
        
    # Create final dataframe
    titer_vs_age_df = pd.DataFrame(
        titer_vs_distance_comparisons,
        columns=[
            'virus_comparison',
            'r_squared',
            'ectodomain_dist',
            'HA1_dist',
            'cohort_comparison'
        ]
    )

    return titer_vs_age_df
```

```python
df = all_titers_2023circulating_only
df = df[~df['timepoint'].isin(timepoints_to_exclude)]
df = df.query('serum != "SCH23_y2021_s057"')
unique_viruses = df.virus.unique()
df = df.copy() 

circulating_titer_vs_distance_df = get_correlations_titers_vs_distance(df, unique_viruses)
circulating_titer_vs_distance_df
```

```python
# Plot correlation coefficients versus difference in ages
use_sequence = 'HA1'
cohorts_to_plot = ['SCH', 'PennVaccineCohort', 'PennVaccineCohort_SCH']
longer_label_dict = {'SCH': 'children (n=56)', 
                     'PennVaccineCohort': 'pre-vaccination adults (n=39)',
                     'PennVaccineCohort_SCH': 'adults and children (n=95)'}
df = (circulating_titer_vs_distance_df[circulating_titer_vs_distance_df['cohort_comparison'].isin(cohorts_to_plot)]
      .replace(longer_label_dict)
     )


# Configure simple scatter
fill = True
opacity = 0.15
stroke = 'black'
strokeWidth = 1.2
markSize = 120
titleFontSize = 16
labelFontSize = 16
RtextLocation = 90
width = 200
height = width
_range = [60, 30000]


charts = []

for cohort in [longer_label_dict.get(item, item) for item in cohorts_to_plot]:
    sub_df = df[df['cohort_comparison'] == cohort]

    # Fit R²
    if len(sub_df) > 1:
        model = LinearRegression()
        x = sub_df[[f'{use_sequence}_dist']]
        y = sub_df[['r_squared']]

        # Implement scaling
        scaler = StandardScaler()
        x_scaled = scaler.fit_transform(x)
        y_scaled = scaler.fit_transform(y)

        # Fit model and calculate R2
        model.fit(x_scaled, y_scaled.ravel())
        # r2 = model.score(x_scaled, y_scaled)
        # r2_string = f"R² = {r2:.2f}"

        # Predict
        y_pred = model.predict(x_scaled)
        # Compute R (Pearson correlation between predicted and actual values)
        print(f"Slope: {model.coef_[0]:.2f}")
        print(f"R: {np.corrcoef(y_scaled.ravel(), y_pred.ravel())[0, 1]}")
        r_string = f"R = {model.coef_[0]:.2f}"

    else:
        r_string = ""

    # Determine annotation position
    x_pos = sub_df[f'{use_sequence}_dist'].max() * 0.9
    y_pos = sub_df['r_squared'].max() * 0.9

    # Scatter chart
    scatter = alt.Chart(sub_df).mark_circle(
        size=markSize,
        stroke=stroke,
        strokeWidth=strokeWidth,
        filled=fill,
        opacity=opacity
    ).encode(
        x=alt.X(f'{use_sequence}_dist:Q',
                title=['Hamming distance', 'between pairs of viruses'],
                axis=alt.Axis(grid=False, titleFontSize=titleFontSize, labelFontSize=labelFontSize, titleFontWeight='normal')),
        y=alt.Y('r_squared:Q',
                title=['correlation coefficient (R2)', 'between pairs of viruses'],
                scale = alt.Scale(domain=[0,1]),
                axis=alt.Axis(grid=False, titleFontSize=titleFontSize, labelFontSize=labelFontSize, titleFontWeight='normal')),
    ).properties(
        width=width,
        height=height,
        title=alt.TitleParams(text=cohort, fontSize=titleFontSize, anchor='middle', fontWeight='normal')
    )

    # Text annotation
    text = alt.Chart(pd.DataFrame({
        'label': [r_string]
    })).mark_text(
        align='right',
        baseline='top',
        dx=RtextLocation, dy=-RtextLocation,
        fontSize=labelFontSize,
        color='black'
    ).encode(
        text='label:N'
    )

    # Combine and append
    charts.append(scatter + text)

# --- Concatenate all charts horizontally ---
final_plot = alt.hconcat(*charts).configure_axis(grid=False).properties(
    title=alt.TitleParams(
        text='titer correlations vs Hamming distance among 2023-circulating strains', 
        anchor='middle', 
        fontSize=18,
    )
)
final_plot

# Save plot
outfile = os.path.join(resultsdir, 'titer_vs_Hamming_circulating.svg')
final_plot.save(outfile, width = 1800)

# Display plot
final_plot
```
